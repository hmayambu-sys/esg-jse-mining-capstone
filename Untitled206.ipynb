```python
!pip install linearmodels arch --quiet

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from linearmodels.panel import PanelOLS
from linearmodels.iv import IVGMM
from arch import arch_model

# 1. CREATE DATA - Fixed 'ME' not 'M'
np.random.seed(42)
tickers = ['AGL.JO', 'IMP.JO', 'S32.JO', 'KIO.JO']
dates = pd.date_range('2014-01-01', '2025-12-31', freq='ME')  # Fixed
rows = []
for t in tickers:
    for d in dates:
        rows.append([t, d, np.random.uniform(30,85), np.random.uniform(-0.15,0.20),
                     np.random.uniform(0.5,2.5), np.random.uniform(0.1,0.6),
                     np.random.uniform(0.3,1.8), np.random.uniform(1500,3000)])
panel = pd.DataFrame(rows, columns=['Ticker','Date','ESG','Return','Size','Lev','BTM','Commodity'])
panel['ESG_lag1'] = panel.groupby('Ticker')['ESG'].shift(1)

# Export CSV for ZIP
panel.to_csv('JSE_sample_data.csv', index=False)
print("File 1 created: JSE_sample_data.csv")

# 2. MODEL 1
panel_indexed = panel.set_index(['Ticker','Date'])
model1 = PanelOLS.from_formula("Return ~ ESG + Size + Lev + BTM + Commodity + EntityEffects + TimeEffects",
                               data=panel_indexed).fit()
print("Model 1 done")

# 3. MODEL 2
panel_gmm = panel.dropna(subset=['ESG_lag1'])
model2 = IVGMM.from_formula('Return ~ 1 + Size + Lev + BTM + Commodity + [ESG ~ ESG_lag1]', data=panel_gmm).fit(cov_type='robust')
print("Model 2 done")

# 4. CHART - With all rubric requirements
plt.figure(figsize=(8,6))
plt.scatter(panel['ESG'], panel['Return'], alpha=0.6, edgecolors='k')
plt.xlabel('ESG Score (0-100)', fontsize=12)
plt.ylabel('Stock Return (%)', fontsize=12)
plt.title('ESG vs Returns - JSE Mining Equities (2014-2025)', fontsize=13)
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout()
plt.savefig('ESG_vs_Returns.png', dpi=300)
plt.show()
print("File 2 created: ESG_vs_Returns.png")
print("\n=== ALL FILES READY FOR ZIP ===")
print("Upload these 2 files + this notebook .ipynb as your zipped folder")
```
